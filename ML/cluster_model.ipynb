```python
from sklearn.datasets import load_iris
import pandas as pd
import numpy as np

iris = load_iris()

df = pd.DataFrame(data=iris.data, columns=iris.feature_names)

df
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import accuracy_score

km = KMeans(n_clusters=3, random_state=42)

km.fit(df)

km.labels_

accuracy_score(km.labels_, iris.target)
```

```python
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

pca_df = pca.fit_transform(df)

pca_df
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.scatter(pca_df[:, 0], pca_df[:, 1], c=km.labels_, cmap='viridis')
plt.title('Iris Dataset - K-Means Clustering')
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.grid()
plt.show()
```

```python
from sklearn.cluster import DBSCAN

ds = DBSCAN(eps=0.5, min_samples=5)

labels = ds.fit_predict(df)

labels
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.scatter(pca_df[:, 0], pca_df[:, 1], c=labels, cmap='viridis')
plt.title('Iris Dataset - DBSCAN Clustering')
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.grid()
plt.show()
```
