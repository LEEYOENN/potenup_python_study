```python
movie = pd.read_csv("../../data_resources/movie/movies.csv")
rating = pd.read_csv("../../data_resources/movie/ratings.csv")
```

```python
df = pd.merge(movie, rating, on="movieId")

df
```

```python
title_user = df.pivot_table(index="userId", columns="title", values="rating")
title_user
title_user.fillna(0, inplace=True)

rating_matrix = title_user.T

rating_matrix
```

```python
from sklearn.metrics.pairwise import cosine_similarity
item_sim = cosine_similarity(rating_matrix, rating_matrix)

item_sim_df = pd.DataFrame(data=item_sim, index=rating_matrix.index, columns=rating_matrix.index)
item_sim_df
```

```python
movies = pd.read_csv("../../data_resources/movie/movies.csv")
movies.head()
ratings = pd.read_csv("../../data_resources/movie/ratings.csv")
ratings.head()
```

```python
user9_movies = set(ratings[ratings["userId"] == 9]['movieId'])
user9_movies
whole_movie = set(movies['movieId'])

whole_movie - user9_movies
```

```python
def get_unseen_movies(userId):
    return list(set(movies['movieId']) - set(ratings[ratings['userId'] == userId]['movieId']))

unseen_movies = get_unseen_movies(34)
len(unseen_movies)
```

```python
pred_result = []
userId = 34
for movieId in unseen_movies:
    result = svd.predict( str(userId), str(movieId) )

    temp_dict = {}
    temp_dict["id"] = result.iid
    temp_dict["est_rating"] = result.est
    temp_dict["title"] = movies[movies['movieId'] == int(result.iid)]['title'].iloc[0]

    pred_result.append(temp_dict)

result_df = pd.DataFrame(pred_result)
result_df.sort_values(by="est_rating", ascending=False, inplace=True)
# pred_result.sort(key=lambda x: x['est_rating'], reverse=True)
# pred_result[:10]
result_df[:5]
```
