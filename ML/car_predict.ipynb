```python
import pandas as pd
import numpy as np
import warnings

```

```python
df = pd.read_csv("../../data_resources/CarPrice_Assignment.csv")
pd.set_option('display.max_columns', None)
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
def labeling_columns_simple(data_frame, columns):

    for column in columns:
        data_frame[column] = LabelEncoder().fit_transform(data_frame[column])
```

```python
print(df.info())
```

```python
print(df.aspiration.unique())
print(df.CarName.unique())
print(df.doornumber.unique())
print(df.carbody.unique())
print(df.drivewheel.unique())
print(df.enginelocation.unique())
print(df.enginetype.unique())
print(df.cylindernumber.unique())
print(df.fuelsystem.unique())
```

```python
df['brand'] = df["CarName"].apply(lambda x: x.split(' ')[0].lower())

df.drop("CarName", axis=1, inplace=True)
```

```python
labeling_columns_simple(df, ['aspiration', 'brand', 'fueltype', 'doornumber', 'carbody', 'drivewheel', 'enginelocation', 'enginetype', 'cylindernumber', 'fuelsystem'])
print(df.info())
```

```python
df_raw = df.copy()

df.drop("car_ID", axis=1, inplace=True)
```

```python
df
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error

warnings.filterwarnings(action='ignore')

x = df.drop("price", axis=1)
y = df["price"]

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

rf = RandomForestRegressor(n_estimators=3000, random_state=42)
rf.fit(x_train, y_train)

y_pred = rf.predict(x_test)
 
print("RandomForest:", rf.score(x_test, y_test))
print("r2 score:", r2_score(y_test, y_pred))
print("mean_squared_error:", mean_squared_error(y_test, y_pred))
print("mean_absolute_error:", mean_absolute_error(y_test, y_pred))
```

```python
from catboost import CatBoostRegressor

cat = CatBoostRegressor(verbose=0, random_state=42)
cat.fit(x_train, y_train)
print("CatBoost:", cat.score(x_test, y_test))
```

```python
from sklearn.model_selection import RandomizedSearchCV
from catboost import CatBoostRegressor
from scipy.stats import randint, uniform

# 하이퍼파라미터 탐색 공간 정의
param_dist = {
    'iterations': randint(300, 1000),
    'learning_rate': uniform(0.01, 0.2),
    'depth': randint(4, 10),
    'l2_leaf_reg': uniform(1, 10),
    'bagging_temperature': uniform(0, 1),
    'random_strength': uniform(1, 20),
    'border_count': randint(32, 128)
}

cat = CatBoostRegressor(verbose=0, random_state=42)

# 랜덤 탐색 정의
random_search = RandomizedSearchCV(
    estimator=cat,
    param_distributions=param_dist,
    n_iter=30,
    scoring='r2',
    cv=3,
    verbose=1,
    n_jobs=-1,
    random_state=42
)

# 튜닝 시작
random_search.fit(x_train, y_train)

# 결과 출력
print("Best R^2 Score:", random_search.best_score_)
print("Best Params:", random_search.best_params_)
```
