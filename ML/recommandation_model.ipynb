## 칼럼을 벡터화

```python
name_result = []

for genres in movies["genres"] :
    temp_list = []
    for genre in genres:
        temp_list.append(genre["name"])
    name_result.append(temp_list)

name_result
    
```

```python
movies["genres"] = name_result
```

```python
keyword_result = []

for keywords in movies["keywords"] :
    temp_list = []
    for keyword in keywords:
        temp_list.append(keyword["name"])
    keyword_result.append(temp_list)

movies["keywords"] = keyword_result
```

```python
movies.head()
```

### CountVectorizer 사용

```python
data_result = []

for genres in movies["genres"]:
    data = " ".join(genres)
    data_result.append(data)

movies["genres_literal"] = data_result

movies.genres_literal
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(min_df=1, ngram_range=(1,2)) #전체 문서에서 최소 한번 등장한 단어 추출, 한단어씩 먼저 만들고, 두개씩 묶어서도 만들어라 -> 한번 이상 등장해야 count

genres_vect = cv.fit_transform(movies['genres_literal'])

cv.vocabulary_
```

```python
print(genres_vect.shape)

print(genres_vect.toarray())

genre_vect_array = genres_vect.toarray()
```

```python
from sklearn.metrics.pairwise import cosine_similarity

genre_sim = cosine_similarity(genre_vect_array, genre_vect_array)

genre_sim_sorted = genre_sim.argsort(axis=1)[:, ::-1]
```

```python
genre_sim
```

```python
genre_sim_sorted[:10, :]
```

```python
def find_sim_movie(df , sorted_index, title_name, top_n=10):
    
    title_movie = df[df['title']==title_name]
    title_index = title_movie.index.values
    print(title_index)

    similar_index = sorted_index[title_index, : (top_n)]
    similar_index = similar_index.reshape(-1)

    return df.iloc[similar_index]
```

```python
movie_result = find_sim_movie(movies, genre_sim_sorted, "Spectre", 10)
movie_result.title
```
