# 경사하강법(Gradient Descent) 실습

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.random.rand(100)
y = x * 0.7
x

plt.figure(figsize=(10,5))
plt.scatter(x,y)
plt.show()
```

```python
x
```

```python
y
```

```python
w = np.random.uniform(-1, 1)
w
```

```python
b = np.random.uniform(-1, 1)
b
```

```python
lr = 0.5 #optimazer - learning rate만 사용

for epoch in range(300):
    y_pred = w*x + b

    error = np.abs(y - y_pred).mean() #MAE

    #미분값 구하기
    dL_dw = ((y_pred - y) * x).mean()
    dL_db = (y_pred - y).mean()

    w = w - lr * dL_dw
    b = b - lr * dL_db

    print("error:", error)

    if epoch % 10 == 0:
        y_pred = w*x + b
        plt.figure(figsize=(8,6))
        plt.scatter(x, y, c="red")
        plt.scatter(x, y_pred, c="blue")
        plt.show()

    if error < 0.001:
        break
    
    
```
