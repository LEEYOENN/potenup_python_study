# 머신러닝·딥러닝 총정리 슬라이드 목차

## 1. 서론
- 빅데이터 모델링의 흐름
- 통계 → 머신러닝 → 딥러닝으로의 발전 IDEA
- 강의 로드맵

## 2. 통계 기반 모델링
- 통계적 가정(정규성, 독립성, 등분산성)
- 회귀분석, 분산분석
- 과적합 vs 과소적합

## 3. 머신러닝
- 머신러닝의 등장 배경 (데이터 기반 학습)
- 지도학습 vs 비지도학습
- 태스크별 알고리즘
  - 예측: 선형회귀, SVM 회귀, 의사결정나무
  - 분류: 로지스틱 회귀, SVM, KNN, Random Forest, XGBoost
  - 군집: K-means, 계층적 군집, DBSCAN
  - 차원축소: PCA, t-SNE, LDA
- 주요 알고리즘: CART, 앙상블 (Bagging vs Boosting)
- 피처 엔지니어링 (Feature Engineering)
  - 전처리 (결측치, 스케일링, 불균형 데이터 처리)
  - 차원 축소 (PCA, t-SNE)
  - 다중공선성과 변수 선택
  - 변수 생성 (Feature Creation)
- 파라미터 vs 하이퍼파라미터

## 4. 딥러닝
- 딥러닝의 등장 배경 (Feature Engineering 자동화)
- 인공신경망(ANN) 구조: 가중치, 편향, Affine, 활성화 함수
- 딥러닝 학습 과정
  - 미니배치 → 순전파 → 손실 계산 → 역전파 → 가중치 업데이트
- 딥러닝 학습의 어려움과 해결책
  - 과적합 (Dropout, 정규화, Early Stopping, Data Augmentation)
  - Gradient Vanishing → ReLU, BatchNorm
- PyTorch 학습 하이퍼파라미터
  - lr, batch size, epoch
  - Optimizer (SGD, Adam)
  - Weight Decay, Dropout
  - Early Stopping, LR Scheduler
- 모델 발전사
  - ANN → CNN (Convolution, Padding, Pooling)
  - RNN (Hidden state, 순차 구조)
  - LSTM (게이트 구조, 장기 의존성 해결)
  - Seq2Seq (입출력 시퀀스 변환)
  - Attention (중요 단어 집중)
  - Self-Attention → Transformer (병렬 학습, 문맥 이해)
  - BERT(인코더), GPT(디코더)
- 응용 전략
  - 전이학습 vs 파인튜닝
  - 파라미터 효율적 파인튜닝 (LoRA, Adapter)

## 5. 정리 & 기출 키워드 체크
- 지도 vs 비지도
- 파라미터 vs 하이퍼파라미터
- 데이터 불균형 처리
- 스케일링 (정규화, 표준화)
- PCA, 다중공선성
- ANN 구조와 활성화 함수 역할
- 딥러닝 학습 과정 (순전파–역전파)
- Optimizer, Gradient Vanishing 해결책
- CNN, RNN, LSTM 핵심 키워드
- Attention, Transformer
- 전이학습 vs 파인튜닝
