## 보스턴 데이터로 딥러닝 모델 설계

```python
import pandas as pd

df = pd.read_csv("../../data_resources/boston.csv")
df
```

```python
df.info()
```

```python
x_train = df.drop(["MEDV"], axis=1).values #numpy 배열로
# x_train = torch.tensor(x_train) #numpy 배열을 tensor로
x_train = torch.FloatTensor(x_train) # tensor로 바꾸면서 자동으로 float로
```

```python
x_train.shape
```

```python
y_train = torch.FloatTensor(df.iloc[:, -1].values)
y_train.shape
```

```python
import torch.nn as nn
from torch.optim.adam import Adam

model = nn.Sequential(
    nn.Linear(13, 100),
    nn.ReLU(),
    nn.Linear(100, 50),
    nn.ReLU(),
    nn.Linear(50, 1)
)

model
```

```python
df.shape
```

```python
model.parameters
```

```python
for param in model.parameters():
    print(param)
```

```python
#최적화 함수
optim = Adam(model.parameters(), lr=0.001)
#몇번 반복할지
epochs = 200
#손실함수
criterion = nn.MSELoss()

for epoch in range(epochs):
    optim.zero_grad() #앞에서 계산했던 기울기를 초기화 하도록

    y_pred = model(x_train)  # 모델 학습시키고 예측값 뽑기   
    loss = criterion(y_pred, y_train)
    #backpropagation
    loss.backward()
    optim.step()

    print("loss:", loss.item())
```

```python
model.state_dict()
```

## GPU 버전

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

#데이터와 모델을 모두 gpu로 올린다
x_train = x_train.to(device)
y_train = y_train.to(device)
model = model.to(device)

#최적화 함수
optim = Adam(model.parameters(), lr=0.001)
#몇번 반복할지
epochs = 200
#손실함수
criterion = nn.MSELoss()

for epoch in range(epochs):
    #앞에서 계산했던 기울기를 0으로 초기화 하도록
    optim.zero_grad() 

    # 모델 학습시키고 예측값 뽑기
    y_pred = model(x_train)
    # 오차값 뽑기  
    loss = criterion(y_pred, y_train)
    
    #오차를 이용해서 backpropagation
    loss.backward()

    #optimizer 이용해서 최적의 가중치 업데이트
    optim.step()

    print("loss:", loss.item())
```
