# ChatPromptTemplate
- 대화 목록 스타일로 프롬프트를 사용
- 튜플 형식으로 교대로, 차례대로 사용, (role, message) 형태로 구성
- role
    - system
    - user(human)
    - ai, assistant

```python
# LangSmith 추적 설정 부분
from dotenv import load_dotenv
import os

load_dotenv()

project_name = "wanted_2nd_prompt_basic"
os.environ["LANGSMITH_PROJECT"] = project_name
```

```python
from langchain_openai import ChatOpenAI

#--- 모델 설정 ---#
model = ChatOpenAI(
    temperature=0.1,
    model="gpt-4.1-mini",
    verbose=True
)
```

```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt = ChatPromptTemplate.from_template(
    "{term} 이 무엇인지 초등학생도 알 수 있게 설명해주세요."
)
chat_prompt
```

```python
chain = chat_prompt | model

result = chain.invoke({"term":"LangChain"})
print(result.content)
```

### chat prompt template 중요함

```python
chat_prompt = ChatPromptTemplate.from_messages(
    template_format=[
        ("system", "당신의 MBTI 는 {mbti} 인, {style} 스타일의 {gender}입니다. 당신의 스타일에 맞게 답변해주세요."),
        ("human", "안녕하세요"),
        ("ai", "안녕하세요, 연애 상담 해드릴까요?"),
        ("human", "{question}에 대해 너의 스타일대로 답변해줘.")
    ]
)
chat_prompt
```

```python
chain = chat_prompt | model
chain
```

```python
result = chain.invoke({
    "gender": "남자",
    "mbti": "ENTP",
    "style": "테토",
    "question": "이상형 알려주세요"
})
```

```python
print(result.content)
```
