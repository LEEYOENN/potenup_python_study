# 프롬프트 고도화

```python
# LangSmith 추적 설정 부분
from dotenv import load_dotenv
import os

load_dotenv()

project_name = "wanted_2nd_prompt_basic"
os.environ["LANGSMITH_PROJECT"] = project_name
```

```python
from langchain_openai import ChatOpenAI

#--- 모델 설정 ---#
model = ChatOpenAI(
    temperature=0.1,
    model="gpt-4.1-mini",
    verbose=True
)
```

```python
from langchain_core.prompts import PromptTemplate

#--- 프롬프트 템플릿 설정 ---#
template = """
최근 {year}년 간 {field} 분야의 기술 발전에 대해서 설명해줘
"""
prompt_template = PromptTemplate.from_template(template)
prompt_template
```

```python
chain = prompt_template | model
response = chain.invoke({"year":3, "field":"AI"})
print(response)
```

### 프롬프트에 고정값을 주고 싶을때 input_variables, partial_variable 인자를 사용할 수 있다.

```python
# field 값은 fix 하겠다 year만 입력 받겠다.

prompt_template_fixed = PromptTemplate(
    template=template,
    input_variables=["year"],
    partial_variables={
        "field":"AI"
    }
)

prompt_template_fixed # 고정값 input variables를 만들고싶으면 PromptTemplate에서만 할 수 있나요?
```

```python
fixed_chain = prompt_template_fixed | model
result = fixed_chain.invoke({"year":3})

print(result.content)
```

### Partial_variable 확장 시켜 사용해보기

```python
from datetime import datetime
```

```python
# 현재 시간 출력 - llm 한테 질문할 때 현재 시각이 input으로 들어가야할 때 사용할 수 있다.
datetime.now().strftime("%Y-%m-%d %H:%M:%S")
```

```python
# 날짜를 변환하는 함수
def get_today():
    return datetime.now().strftime("%Y-%m-%d %H:%M:%S")
```

```python
# 예
template = "현재 시간은 {today} 입니다. 이 시간대에 사람들이 가장 많이 하는 활동을 {number}가지 알려줘."
prompt_template = PromptTemplate(
    template=template,
    input_variables=["number"],
    partial_variables={
        "today": get_today
    }
)
```

```python
prompt_template
```

```python
chain = prompt_template | model
result = chain.invoke({"number":3})
print(result.content)
```

### 프롬프트 템플릿 저장 및 로드

```python
template = "현재 시간은 {today} 입니다. 이 시간대에 사람들이 가장 많이 하는 활동을 {number}가지 알려줘."
prompt_template = PromptTemplate(
    template=template,
)
```
