# Langchain 기본 실습

## 토큰 비용 체크 - LangSmith - Set up tracing

### 랭스미스 추적 설정

```python
# LangSmith 추적 설정 부분
from dotenv import load_dotenv
import os

load_dotenv()

project_name = "wanted_2nd_langchain_basic"
os.environ["LANGSMITH_PROJECT"] = project_name
```

```python
from langchain_openai import ChatOpenAI

model = ChatOpenAI(
    temperature=0.1,
    model="gpt-4.1-mini",
    verbose=True
)
```

### 연결 확인

```python
question = "1년에 한국에 오는 관광객 수는 몇 명인가요?"

# 모델의 답을 얻을 때 invoke 메서드
response = model.invoke(question)
```

```python
print(response.content)
```

```python
# 응답 메타데이터(데이터에 대한 데이터)
response.response_metadata
```

### 스트리밍 출력

```python
answer = model.stream(
    question
)

for token in answer:
    # flush 버퍼를 즉시 비워서 화면에 바로바로 출력하도록 하는 역할
    print(token.content, end="", flush=True)
```
