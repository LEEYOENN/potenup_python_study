1. 프롬프트 템플릿 정하기

```python
# LangSmith 추적 설정 부분
from dotenv import load_dotenv
import os

load_dotenv()

project_name = "wanted_2nd_langchain_basic"
os.environ["LANGSMITH_PROJECT"] = project_name
```

### 체인 한 번에 만들기

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI

#--- 프롬프트 템플릿 설정 ---#
template = """
최근 {year}년 간 {field} 분야의 기술 발전에 대해서 설명해줘
"""
prompt_template = PromptTemplate.from_template(template)

#--- 모델 설정 ---#
model = ChatOpenAI(
    temperature=0.1,
    model="gpt-4.1-mini",
    verbose=True
)

#--- 아웃풋 파서 설정 ---#
outputparser = StrOutputParser()

#--- 체인 설정 ---#
chain = prompt_template | model | outputparser
chain
```

1. invoke 출력 - 답변 한번에 가져옴
2. stream 출력 - streaming으로 하나씩 가져옴
3. batch 출력 - 여러 질문을 한번에 보낼 때 사용
4. ainvoke, astream 비동기 출력

```python
# 배치 단위 출력 - 질문이 여러개 일 때 사용

response = chain.batch([{"year":3, "field": "AI"}, {"year":3, "field": "드론"}])
print(response[0])
```

### astream, ainvoke <- 비동기 출력

```python
# ainvoke 비동기 출력

async_answer = chain.ainvoke({"year":3, "field": "드론"})
```

```python
result = await async_answer
result
```
