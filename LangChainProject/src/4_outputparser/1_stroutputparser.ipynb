# StrOutputParser 공부

```python
# LangSmith 추적 설정 부분
from dotenv import load_dotenv
import os

load_dotenv()

project_name = "wanted_2nd_prompt_basic"
os.environ["LANGSMITH_PROJECT"] = project_name
```

```python
from langchain_openai import ChatOpenAI

#--- 모델 설정 ---#
model = ChatOpenAI(
    temperature=0.1,
    model="gpt-4.1-mini",
    verbose=True
)
```

```python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatMessagePromptTemplate, ChatPromptTemplate

chat_prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "당신의 MBTI 는 {mbti} 인, {style} 스타일의 {gender}입니다. 당신의 스타일에 맞게 답변해주세요."),
        ("human", "안녕하세요"),
        ("ai", "안녕하세요, 연애 상담 해드릴까요?"),
        ("human", "{question}에 대해 너의 스타일대로 답변해줘.")
    ]
)
chat_prompt
```

```python
from langchain_core.prompts import ChatMessagePromptTemplate, ChatPromptTemplate

chat_prompt = ChatMessagePromptTemplate.from_template(
    template=[
        ("system", "당신의 MBTI 는 {mbti} 인, {style} 스타일의 {gender}입니다. 당신의 스타일에 맞게 답변해주세요."),
        ("human", "안녕하세요"),
        ("ai", "안녕하세요, 연애 상담 해드릴까요?"),
        ("human", "{question}에 대해 너의 스타일대로 답변해줘.")
    ],
    partial_variables={
        "mbti"
    }
)
chat_prompt
```

```python
chain = chat_prompt | model | StrOutputParser()
chain
```

```python
result = chain.invoke({
    'gender': "여자",
    "mbti": "ENTJ",
    "question": "너랑 잘 맞는 mbti 알려줄 수 있어?",
    "style": "에겐"
})
```

```python
# StrOutputParser 가 붙어서 문자열만 나옴
print(result)
```
