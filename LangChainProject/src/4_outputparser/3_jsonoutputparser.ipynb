# Json Output Parser

```python
# LangSmith 추적 설정 부분
from dotenv import load_dotenv
import os

load_dotenv()

project_name = "wanted_2nd_prompt_basic"
os.environ["LANGSMITH_PROJECT"] = project_name
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

#--- 모델 설정 ---#
model = ChatOpenAI(
    temperature=0.1,
    model="gpt-4.1-mini",
    verbose=True
)
```

```python
from langchain_core.output_parsers import JsonOutputParser

json_parser = JsonOutputParser()

template = """
    오직 Json의 문자열만 출력을 해라. 코드블록 금지
    key: {{"title": str, "bullets": list[str], "tone": "neutral | positive | negative"}}
    {text}
"""

prompt_template = ChatPromptTemplate.from_template(
    template=template,

)

chain = prompt_template | model | json_parser
chain
```

```python
result =chain.invoke({"text": "재택근무는 많은 직장인에게 새로운 표준으로 자리 잡았습니다. 가장 큰 장점은 출퇴근에 소요되는 시간을 절약하여 자기 계발이나 여가 등 개인적인 시간을 더 많이 확보할 수 있다는 점입니다. 또한, 기업 입장에서는 사무실 임대료와 같은 고정 비용을 줄일 수 있어 재정적으로도 긍정적인 효과를 기대할 수 있습니다. 이처럼 유연한 근무 환경은 직원들의 직무 만족도를 높여 결과적으로 전반적인 생산성 향상에 기여하는 것으로 평가받고 있습니다."})
```
