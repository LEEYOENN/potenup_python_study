- 프롬프트 역할
    - System
    - User
    - Assistant

```python
import ollama

def chat_gemma(user_text):
    response = ollama.chat(
        model = "gemma3:1b",
        messages=[
            {"role": "system", "content": "재미있게 말해주세요"},
            {"role": "user", "content": user_text}
        ]
    )

    return response.message.content
```

```python
while True:
    question = input("입력하세요")
    if question == "bye":
        break
    answer = chat_gemma(question)
    print(f"Question: {question}")
    print(f"Answer: {answer}")
```

# API 활용한 LLM 챗봇 만들기

```python
# curl http://localhost:11434/api/generate -d '{
#   "model": "llama3.2",
#   "prompt": "Why is the sky blue?"
# }'

import requests

request_url = "http://localhost:11434/api/generate"
data = {
    "model": "gemma3:1b",
    "prompt": "Why is the sky blue?"
}

response = requests.post(request_url, json=data)
print(response.text)
```

```python
import json

with requests.post(request_url, json=data) as response:
    for line in response.iter_lines():
        print(json.loads(line)["response"], end="")
```

```python
# Send a chat message with a streaming response.

# curl http://localhost:11434/api/chat -d '{
#   "model": "llama3.2",
#   "messages": [
#     {
#       "role": "user",
#       "content": "why is the sky blue?"
#     }
#   ]
# }'

import requests

request_url = "http://localhost:11434/api/chat"
data = {
    "model": "llama3.2",
    "messages": [
    {
        "role": "user",
        "content": "why is the sky blue?"
    }
    ],
    "stream": False
}
```

```python

response = requests.post(request_url, json=data)
print(response.text)
```

```python
print(json.loads(response.text)["message"]["content"])
print(response.json()["message"]["content"])
```
