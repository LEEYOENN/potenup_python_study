# 3. 응답 받기

```python
# GPT5 모델에서 새롭게 생긴 파라미터는 verbosity: low, medium, high

def chat(system_prompt, user_prompt):
    response = client.chat.completions.create(
        model="gpt-4.1-nano",
        messages=[
            {
                "role": "system",
                "content": system_prompt
            },
            {
                "role": "user",
                "content": user_prompt
            }
        ],
        temperature=0.5
    )

    return response.choices[0].message.content
```

```python
print(chat("당신은 인공지능 전문가입니다.", "Transformer 모델을 설명해주세요."))
```

## 3-1 ChatGPT 5 체험해보기

```python
def chat5(system_prompt, user_prompt, verbosity):
    response = client.chat.completions.create(
        model="gpt-5-nano",
        messages=[
            {
                "role": "system",
                "content": system_prompt
            },
            {
                "role": "user",
                "content": user_prompt
            }
        ],
        verbosity=verbosity
    )

    return response.choices[0].message.content
```

```python
# 답변 생성기
# 리뷰를 읽고, 리뷰에서 긍정적인 부분, 부정적인 부분을 나열한다. - chat1
# 긍/부정에 따라 말투를 설정하고, 이를 기반하여 답변한다.(조건은 부정적인 것 먼저) - chat2
```

```python
# chat1: 리뷰를 입력하면 긍/부정 리스트를 반환한다.
# chat2: 긍/부정 리스트를 받으면 그에 따른 댓글 답변을 작성한다.
```

```python
import openai
import base64
from IPython.display import Image as IPImage, display

def generate_image(prompt):
    #요청 및 응답
    response = openai.images.generate(
        model='gpt-image-1',
        prompt=prompt,
        size="1024x1024",
        quality="auto",
        n=1
    )

    # 이미지 출력
    image_base64 = response.data[0].b64_json
    image_bytes = base64.b64decode(image_base64)
    display(IPImage(data=image_bytes, width=400))
```

```python
# 목표: 우리반 귀여운 마스코트 캐릭터 만들기
prompt = """An extremely cute and fluffy brown rabbit mascot, in a vibrant 3D rendering style, looking intently at a laptop. The rabbit's fur should be highly detailed and soft, and the overall aesthetic should be charming and photorealistic, similar to a character from a high-quality animated movie."""
generate_image(prompt)
```
