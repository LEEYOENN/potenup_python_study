```python
# 환경변수 불러오기
from dotenv import load_dotenv
load_dotenv()

from openai import OpenAI
client = OpenAI()

response = client.chat.completions.create(
    model="gpt-4.1-nano",
    messages=[
        {
            "role": "system",
            "content": "당신은 인공지능 전문가입니다." # 시스템 프롬프트에 모델의 페르소나를 지정
        },
        {
            "role": "user",
            "content": "Transformer 모델을 설명해주세요" # 유저 프롬프트에 응답받고 싶은 내용을 기잉ㅂ
        }
    ]
    temperature=0.5
)

answer = response.choices[0].message.content
print(answer)
```

```python
# 환경변수 불러오기
from dotenv import load_dotenv
load_dotenv()

from openai import OpenAI
client = OpenAI()

response = client.chat.completions.create(
    model="gpt-4.1-nano",
    messages=[
        {
            "role": "system",
            "content": "김한호강사님은 원티드랩이라는 부트캠프의 강사 김한호입니다. 학생들에게 말하듯이 존댓말로 친근하게 말해주세요. 자신의 이름도 한번 언급하고 응답을 작성해주세요"  # 시스템 프롬프트에 모델의 페르소나를 지정
        },
        {
            "role": "user",
            "content": "친구에게 전하는 응원 메세지를 작성해주세요." # 유저 프롬프트에 응답받고 싶은 내용을 기입
        }
    ],
    temperature=0.5
)

answer = response.choices[0].message.content
print(answer)
```
