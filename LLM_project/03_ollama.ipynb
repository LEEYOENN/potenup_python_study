# 미니 프로젝트: Ollama로 페르소나가 부여된 챗봇 만들기

```python
# 1. llama가 아닌 다른 모델 다운로드 받아보기 
    # - deepseek-r1:8b 모델을 다운받아서 로컬에서 deepseek LLM을 사용해보았습니다.

# 2. OpenAPI KEY를 쓸 때와 로컬 LLM을 쓸 때의 차이점 비교하기 
    # - 우선 환경변수에서 API KEY를 불러올 필요가 없고 다운로드한 모델명만 알려주면 되어 간편했습니다.
    # - deepseek-r1:8b 모델은 사고 과정을 content안에 <think /> annotation을 함께 반환합니다.
# 3. 속도 비교하기
    # gpt-4.1-nano 모델을 API 로 불러와 사용했을 때는 응답 당 약 5.5초가 걸렸으나
    # deepseek-r1:8b를 로컬로 사용했을 때는 응답 당 약 20초가 걸렸습니다.
    # llama3.1:8b 모델을 로컬로 사용했을 때는 응답이 약 6초가 걸렸습니다.
    # deepseek 모델이 상대적으로 응답이 오래걸리는 것을 확인
```

**사용모델1: deepseek-r1:8b**
- d-eepseek-r1:8b 모델을  로컬로 사용했을 때는 응답 당 약 20초 가량 걸렸습니다.
- deepseek-r1:8b 모델은 사고 과정을 content안에 <think /> annotation 안에 함께 반환합니다.
- <think /> annotation 안에 내용이 실제 답변보다 훨씬 깁니다.

**사용모델2: gemma3:1b**
- gemma3:1b 모델을 로컬로 사용했을 때는 응답이 약 10초가 걸렸습니다.
- 답변 퀄리티 좋습니다.

**사용모델3: llama3.1:8b**
- llama3.1:8b 모델을 로컬로 사용했을 때는 응답이 약 6초가 걸렸습니다.
- 자기를 라린이라고 소개했는데 그게 뭔지 모르겠습니다.

```python
import ollama

# LLM 리스트 조회
local_llm_list = ollama.list()
for llm in local_llm_list:
    print(llm[1])
```

```python
import ollama

def chat(system_prompt, user):
    response = ollama.chat(
    model="deepseek-r1:8b",
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user}
    ]
    )

    return response.message.content
```

```python
system_prompt = "당신은 AI에이전트 선생님입니다. 학생에게 설명한다고 생각하고 질문을 이해하기 쉽게 설명해주세요"
user = "나 langchain 사용법이 궁금해"
```

```python
response = chat(system_prompt, user)
print(response)
```

```python
while True:
    user = input(">")

    if (user == "그만"):
        break
    
    response = chat(system_prompt, user)
    print(response)
    
```

```python
import ollama

def chat_gemma(system_prompt, user):
    response = ollama.chat(
    model="gemma3:1b",
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user}
    ]
    )

    return response.message.content
```
