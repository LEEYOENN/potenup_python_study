## RNN (Recurrent Neural Network)

- 입력을 순차적으로 처리

- 이전 단계(hidden state)의 출력을 다음 단계 입력과 함께 사용

**한계:**

- 시계열 정보를 기억할 수는 있지만, **단기 기억(short-term memory)**에 국한됨.

- 시퀀스가 길어지면 기울기 소실(Vanishing Gradient) 문제 때문에 앞쪽의 정보가 뒷부분에 잘 전달되지 않음.

- 즉, 오래된 정보는 금방 잊어버리고 최근 정보만 반영되는 경향

## LSTM (Long Short-Term Memory)

- RNN의 단기 기억 한계를 극복하기 위해 등장

- 핵심 아이디어: Cell State(셀 상태) 라는 “정보를 오래 유지하는 저장 공간”을 두고, 정보의 흐름을 **Gate(게이트)**로 조절

#### 3가지 게이트:

- Forget Gate (망각 게이트)

    - 어떤 정보를 버릴지 결정.

    - “이전 입력에서 의미 없는 값은 지우자.”

- Input Gate (입력 게이트)

    - 새로 들어오는 정보를 얼마나 저장할지 결정

    - “이 단어는 중요한 의미를 가지니 저장하자.”

- Output Gate (출력 게이트)

    - 현재 시점에서 어떤 정보를 출력(hidden state)할지 결정

    - “지금 이 단어를 예측하는 데 필요한 정보는 이 부분이다.”


- 장기 의존성(Long-term dependency)을 학습 가능

- 긴 문장, 시계열 데이터에서도 맥락을 잘 반영

- 기울기 소실 문제를 크게 완화

## Seq2Seq 기본 개념

- 목적: 입력 시퀀스를 받아 다른 형태의 출력 시퀀스를 생성하는 모델.

- 예: 번역(영어 → 한국어), 챗봇(질문 → 답변), 요약(긴 문장 → 짧은 요약)

#### 구조
1. **인코더 (Encoder)**

- 입력 시퀀스를 순차적으로 읽어 hidden state를 업데이트

- 마지막 시점에서 나온 hidden state들을 압축해서 하나의 **Context Vector(문맥 벡터)**로 만듦

- 이 컨텍스트 벡터는 “입력 문장의 전체 의미”를 담고 있다고 볼 수 있음

2. **디코더 (Decoder)**

- 인코더가 만든 컨텍스트 벡터를 입력으로 받아, step-by-step으로 출력을 생성

- 보통 이전 시점의 출력(토큰)도 함께 입력으로 사용해서 다음 단어를 예측

**한계와 개선**

- 기본 Seq2Seq의 한계: 긴 문장을 인코더의 마지막 컨텍스트 벡터 하나에만 압축해서 담다 보니, 정보 손실이 발생.

- 개선 방법:

    - Attention Mechanism (Bahdanau, Luong) → 단일 벡터 대신, 인코더의 모든 hidden state에 “가중치”를 주어 필요한 부분을 동적으로 참조.

    - 이게 나중에 Transformer 구조로 발전.

## Attention

- 모든 입력이 같은 중요도를 가지지 않는다라는 개념에서 출발

- 입력 시퀀스의 모든 위치를 동시에 고려하기 때문에 병렬처리 가능

    - 쿼리: 모델이 현재 주목하고있는 정보
    - 키: 모든 입력에 대한 기준을 제공하는 정보
    - 값: 각 입력에 대한 출력

- 쿼리와 키의 유사도를 계산해서 중요도 점수를 계산

- 유사도를 기반으로 가중치(Softmax) 계산

- 가중치를 값에 곱하고 최종 출력

## 셀프 어텐션

- 문장 내부 요소들 간의 중요도를 계산, 키 값 모두 같은 입력 시퀀스에서 계산

## Transformer

- 긴 의존 관계 처리 문제, 병렬처리 제약 문제를 해결하는 모델

#### - Attention is All You Need

- 순차 처리가 아닌 전체적 관계를 한번에 처리하는 방식 제시

- **인코더 디코더** 방식
    - 인코더: 입력 시퀀스를 처리(입력 시퀀스 → 인코더 출력)
    - 디코더: 처리된 정보를 바탕으로 출력 시퀀스 생성

#### 구조
**Encoder**
- Input Embedding + Positional Encoding
    - 각 단어를 벡터로 변환(Embedding)
    - 순서 정보를 담기 위해 Positional Encoding 추가

- Multi-head Self-Attention
    - 모든 문장에 대한 셀프어텐션 수행 층

- Position-wise Feed Foward Neural Network
    - 완전연결신경망 층

- 이 과정을 N번 반복

<br/>

**Decoder**
- Output Embedding + Positional Encoding

- Masked Multi-head Self-Attention 
    - Mask를 이용해 어텐션 계산
    - 아직 보지 않은 단어들의 정보를 차단함

- Position-wise Feed Foward Neural Network
    - 완전연결신경망 층

-  Multi-head Cross-Attention
    - Encoder에서 나온 hidden states와 디코더 입력을 연결
    - 디코더가 단어를 생성할 때, 인코더 문장 전체에서 어떤 부분을 참고할지 결정

- 이 과정을 N번 반복
