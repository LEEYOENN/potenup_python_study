## 2. 이미지 예측해보기

```python
# 이미지 1개 예측하기
results = model("./images/yolo_image.jpg", save=True, project="result") # model() == model.predict() # [이미지 여러개] 넣기 가능
results 
```

```python
for rslt in results:
    boxes = rslt.boxes # 이미지에서 디텍션한 객체들에 대한 정보 담겨있음
    masks = rslt.masks
    keypoints = rslt.keypoints
    probs = rslt.probs
    obb = rslt.obb
    names = rslt.names
    rslt.show()
results[0].names
print(names) # yolo11n 모델이 찾을 수 있는 객체 딕셔너리
    
```

```python
print(type(boxes))
print(boxes.cls) # 예측 클래스
print(boxes.conf) # 신뢰도(confidence)
print(boxes.data) # x1, y1, x2, y2, conf, cls 가 합쳐져서 한 행에 나온거
print(len(boxes.data))
boxes
```

```python
pred_names = []
for idx in boxes.cls:
    pred_names.append(names[int(idx)])
print(pred_names)
```

```python
# x1, y1, x2, y2, conf, cls 가 합쳐져서 한 행에 나온거
print(boxes.data[0])
```

```python
result = {"x1":[], "y1":[], "x2":[], "y2":[], "conf":[], "cls":[]}
for x1, y1, x2, y2, conf, cls in boxes.data: # x1, y1, x2, y2, conf, cls
    print(x1, y1, x2, y2, conf, cls)
    result["x1"].append(x1)
    result["y1"].append(y1)
    result["x2"].append(x2)
    result["y2"].append(y2)
    result["conf"].append(conf)
    result["cls"].append(cls)
print(result)
```
