```python
# 분기형 코드(예: 20251 = 2025년 1분기) 처리
def parse_quarter_str(quarter_str):
    s = str(quarter_str)
    if len(s) == 5:
        year = int(s[:4])
        quarter = int(s[4])
        # 분기 첫 달(1,4,7,10월)로 변환
        month = (quarter - 1) * 3 + 1
        return pd.Timestamp(year=year, month=month, day=1)
    elif len(s) == 6:
        return pd.to_datetime(s, format='%Y%m')
    else:
        return pd.NaT


Commercial_district_data_성수 = data[data['상권_코드_명'].isin(["성수역 골목형상점가","성수역", "성수동카페거리"])]
# Commercial_district_data_성수['기준_년분기_코드'] = Commercial_district_data_성수['기준_년분기_코드'].astype(str).str.zfill(6)

Commercial_district_data_성수['date'] = Commercial_district_data_성수['기준_년분기_코드'].apply(parse_quarter_str)
Commercial_district_data_성수["date"] = pd.to_datetime(Commercial_district_data_성수['date'], format='YYYYM')
Commercial_district_data_성수["year"] = Commercial_district_data_성수['date'].dt.year


Commercial_district_data_성수[['year', "총_유동인구_수"]].groupby(by="year").sum()
```

```python
# 분기형 코드(예: 20251 = 2025년 1분기) 처리
def parse_quarter_str(quarter_str):
    s = str(quarter_str)
    if len(s) == 5:
        year = int(s[:4])
        quarter = int(s[4])
        # 분기 첫 달(1,4,7,10월)로 변환
        month = (quarter - 1) * 3 + 1
        return pd.Timestamp(year=year, month=month, day=1)
    elif len(s) == 6:
        return pd.to_datetime(s, format='%Y%m')
    else:
        return pd.NaT


Commercial_district_data_가로수길 = data[data['상권_코드_명'].str.contains('가로수길', na=False)]
# Commercial_district_data_가로수길['기준_년분기_코드'] = Commercial_district_data_가로수길['기준_년분기_코드'].astype(str).str.zfill(6)

Commercial_district_data_가로수길['date'] = Commercial_district_data_가로수길['기준_년분기_코드'].apply(parse_quarter_str)
Commercial_district_data_가로수길["date"] = pd.to_datetime(Commercial_district_data_가로수길['date'], format='YYYYM')
Commercial_district_data_가로수길["year"] = Commercial_district_data_가로수길['date'].dt.year


Commercial_district_data_가로수길[['year', "총_유동인구_수"]].groupby(by="year").sum()
```

```python
# Commercial_district_data = Commercial_district_data_성수[['year', "총_유동인구_수"]].groupby(by="year").sum()
# Commercial_district_data_가로수길_result = Commercial_district_data_가로수길[['year', "총_유동인구_수"]].groupby(by="year").sum()

Commercial_district_data_성수 = Commercial_district_data_성수.rename(columns={'총_유동인구_수': '성수_총_유동인구_수'})
Commercial_district_data_가로수길 = Commercial_district_data_가로수길.rename(columns={'총_유동인구_수': '가로수길_총_유동인구_수'})

Commercial_district_data = pd.concat([
    Commercial_district_data_성수[['year', "성수_총_유동인구_수"]].groupby(by="year").sum(), 
    Commercial_district_data_가로수길[['year', "가로수길_총_유동인구_수"]].groupby(by="year").sum()
    ],
    axis=1
)
Commercial_district_data.loc[2025] = Commercial_district_data.loc[2025]*4
```
