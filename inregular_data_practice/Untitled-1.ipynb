# Import Required Libraries
pandas 라이브러리를 import 합니다.

```python
import pandas as pd
```

# Load CSV Data for Each Year
2019년부터 2024년까지 각 연도별 CSV 파일을 pandas DataFrame으로 읽어옵니다.

```python
df2019 = pd.read_csv("../../data_resources/2019_서울시_상권분석서비스(점포-상권).csv", encoding='cp949')
df2020 = pd.read_csv("../../data_resources/2020_서울시_상권분석서비스(점포-상권).csv", encoding='cp949')
df2021 = pd.read_csv("../../data_resources/2021_서울시_상권분석서비스(점포-상권).csv", encoding='cp949')
df2022 = pd.read_csv("../../data_resources/2022_서울시_상권분석서비스(점포-상권).csv", encoding='cp949')
df2023 = pd.read_csv("../../data_resources/2023_서울시_상권분석서비스(점포-상권).csv", encoding='cp949')
df2024 = pd.read_csv("../../data_resources/2024_서울시 상권분석서비스(점포-상권).csv", encoding='cp949')
```

# Check and Align DataFrame Columns
각 DataFrame의 columns를 확인하고, 컬럼명이 다를 경우 맞춰줍니다.

```python
# 각 연도별 컬럼 확인
print("2019:", df2019.columns)
print("2020:", df2020.columns)
print("2021:", df2021.columns)
print("2022:", df2022.columns)
print("2023:", df2023.columns)
print("2024:", df2024.columns)

# 공통 컬럼 추출 및 맞추기
common_cols = set(df2019.columns)
for df in [df2020, df2021, df2022, df2023, df2024]:
    common_cols = common_cols & set(df.columns)
common_cols = list(common_cols)

# 모든 DataFrame을 공통 컬럼으로 맞추기
df2019 = df2019[common_cols]
df2020 = df2020[common_cols]
df2021 = df2021[common_cols]
df2022 = df2022[common_cols]
df2023 = df2023[common_cols]
df2024 = df2024[common_cols]
```

# Concatenate DataFrames by Rows
pd.concat()을 사용하여 연도별 DataFrame을 행 기준으로 병합합니다.

```python
merged_df = pd.concat([df2019, df2020, df2021, df2022, df2023, df2024], axis=0, ignore_index=True)
print(merged_df.shape)
merged_df.head()
```

# Save Merged Data to CSV
병합된 DataFrame을 새로운 CSV 파일로 저장합니다.

```python
merged_df.to_csv("../../data_resources/2019_2024_서울시_상권분석서비스_병합.csv", index=False, encoding='cp949')
```
